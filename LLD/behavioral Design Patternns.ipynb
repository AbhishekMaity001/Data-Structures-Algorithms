# Observer Pattern


 Example -> Subscribers subscribing(observing) Youtube channel 

```python
## Create an interface of a subscriber

from abc import ABC, abstractmethod

class Subscriber(ABC):

    @abstractmethod
    def update(self, message):
        pass


class EmailSubscriber(Subscriber):

    def __init__(self, username, email):
        self.username = username
        self.email = email
    
    def update(self, message):
        print(f"Email notification sent to the subscriber: {message}")

class MobileSubscriber(Subscriber):

    def __init__(self, username):
        self.username = username
    
    def update(self, message):
        print(f"Mobile notification sent to the subscriber: {message}")

class BellNotificationSubscriber(Subscriber):

    def __init__(self, username):
        self.username = username
    
    def update(self, message):
        print(f"Ting ting!!! Push notification sent to the subscriber: {message}")
    
```

```python
## create a youtube channel that can be subscribed
class YoutubeChannel:
    def __init__(self, name):
        self.name = name
        self.subscribers = []

    def subscribe(self, subscriber:Subscriber ):
        self.subscribers.append(subscriber)

    def unsubscribe(self,subscriber:Subscriber ):
        self.subscribers.remove(subscriber)

    def upload_video(self, video_title): 
        message = f"New video uploaded - {video_title}"
        print(f"New video uploaded - {video_title}")
        self._notify(message)
    
    def _notify(self, message):
        for subscriber in self.subscribers:
            print(f"Sending notificaton to {subscriber.username} ")
            subscriber.update(message)
```

```python
# creating subscriber

rajan = EmailSubscriber(username = "rajanrox", email = "rajan@gmail.com")
mahajan= MobileSubscriber(username = "mahajan123")


# create channel
codeverra = YoutubeChannel(name = "codeverra")

## subscribe to the channel

codeverra.subscribe(rajan)
codeverra.subscribe(mahajan)


## channel will upload a video

codeverra.upload_video(video_title = "Low level design for beginners")
```

```python
## application in e-commerce

class Observer(ABC):

    @abstractmethod
    def update(self, message):
        pass

class Order:

    def __init__(self, order_id: int):
        self.order_id = order_id
        self.status = "PLACED"
        self.observers = []
        
    def add_observer(self, observer: Observer):
        self.observers.append(observer)

    def remove_observer(self, observer: Observer):
        self.observers.append(observer)

    def _notify(self, message):
        for observer in self.observers:
            observer.update(message)
        
    def update_status(self, new_status):
        self.status = new_status # "preparing to ship", "shipped", "out for delivery", "delivered"

        self._notify(new_status)
        # notification_service.notify(...)
        # warehouse_service.update(...)
        # analytics_service.update(...)





class CustomerNotificationService(Observer): 
    def update(self, message):
        print(f"[Customer Notification] Order  is now '{message}'.")
        
class WarehouseService(Observer): 
    def update(self, message):
        print(f"[Warehouse] Updating inventory for Order .")
        
class AnalyticsService(Observer): 
    def update(self, message):
        print(f"[Analytics] Recording status '{message}' for Order.")
    
```

# Strategy Pattern

Design a navigation application similar to Google Maps.
A user enters a source and destination.
The application supports multiple routing preferences:

    Fastest Route
    Shortest Route
    Avoid Tolls
    Scenic Route

The navigation engine should not know how each route is calculated.
It should simply delegate route calculation to the selected routing algorithm.

 avoid the if-else

     if route == "Fastest":
        ...

     elif route == "Shortest":
        ...
    
     elif route == "Avoid Tolls":
        ...

```python
## define interface for a strategy

class RouteStrategy(ABC):

    @abstractmethod
    def find_route(self, source, destination):
        pass


## concrete class


class WalkingRoute(RouteStrategy):
    def find_route(self, source, destination):
        print(f"Finding walking route from {source} to {destination} !!!!!")
    
class ScenicRoute(RouteStrategy):
    def find_route(self, source, destination):
        print(f"Finding scenic route from {source} to {destination}")

class AvoidTollRoute(RouteStrategy):
    def find_route(self, source, destination):
        print(f"Finding toll-free route from {source} to {destination}")

class ShortestRoute(RouteStrategy):
    def find_route(self, source, destination):
        print(f"Finding shortest route from {source} to {destination}")

class FastestRoute(RouteStrategy):
    def find_route(self, source, destination):
        print(f"Finding fastest route from {source} to {destination}")
        
```

```python
class NavigationSystem:

    def __init__(self, strategy:RouteStrategy):
        self.strategy = strategy

    def navigate(self, source, destination):
        self.strategy.find_route(source, destination)

    
        
```

# Command Pattern
        - The request/action/method call itself becomes a class

### Command Pattern ---> that action itself can be a class

### Template for the command pattern :
    - There are 4 main components that we need to define:
            - receiver ( like television) - act on the command
            - invoker ( remote ) - invoke the command
            - Abstract Command (Command interface) ( TurnOn, TurnOff, IncreaseVolume) - tell me what actions are supported
            - Concrete request / Command - implementation of that action

Design a universal remote control for a smart home.

The remote should be capable of controlling different electronic devices such as televisions, fans, and air conditioners.

Initially, the remote supports the following operations for a television:

    Turn On
    Turn Off
    Increase Volume

The business also informs you that future requirements include:

Undo the last operation.
Record a sequence of button presses (Macro).
Schedule commands to execute later.
Log every button press.
Add support for new devices without modifying the remote control.

The remote should not know how each device performs its operation.


```python
class Television:

    def turn_off(self):
        print("television is off!!")

    def turn_on(self):
        print("television is on!!")

    def increase_volume(self):
        print("television is on!!")


class RemoteControl:

     def turn_on(self, television):
         television.turn_on()

     def turn_off(self, television):
         television.turn_off()

     def volume_up(self, television):
        television.increase_volume()
```

```python
## reciever ( television)

class Television:

    def turn_off(self):
        print("television is off!!")

    def turn_on(self):
        print("television is on!!")

    def increase_volume(self):
        print("Volume is increased!!")

    def mute(self):
        print("Now muted!!!!")


## command interface
class Command(ABC):

    @abstractmethod
    def execute(self):
        pass


## Concrete Commands --> represent the actions/commands as class

# Each command stores the receiver and knows
# which operation should be performed.

class TurnOnCommand:
    def __init__(self, television):
        self.television = television

    def execute(self):
        self.television.turn_on()
        
class TurnOffCommand:

    def __init__(self, television):
        self.television = television

    def execute(self):
        self.television.turn_off()


class IncreaseVolumeCommand:
    def __init__(self, television):
        self.television = television

    def execute(self):
        self.television.increase_volume()

class MuteCommand:
    def __init__(self, television):
        self.television = television

    def execute(self):
        self.television.mute()

## Invoker ---> Remote

class RemoteControl:

    def press_button(self, command:Command):
        command.execute()


```

```python
## client


tv = Television()  # create the reciever

remote = RemoteControl() ## create the invoker


power_on = TurnOnCommand(tv)
power_off = TurnOffCommand(tv)
increase_volume = IncreaseVolumeCommand(tv)
mute = MuteCommand(tv)
```

```python
remote.press_button(power_on)
```

```python
## create a class Chef (receiver)
class Chef:
    def __init__(self, name):
        self.name = name

    def prepare_burger(self): 
        print("chef is preparing burger!!!")

    def prepare_pizza(self):
        print("chef is preparing pizza!!!")

    def prepare_fries(self):
        print("chef is preparing fries!!!")


## command interface

class Order(ABC):
    
    @abstractmethod
    def execute(self):
        pass


# concrete commands
from uuid import uuid4
class BurgerOrder(Order):

    def __init__(self, chef):
        self.chef = chef
        self.name = "burger order"
        self.order_id = str( uuid4())

    def execute(self):
        self.chef.prepare_burger()

class PizzaOrder(Order):

    def __init__(self, chef):
        self.chef = chef
        self.name = "pizza order"
        self.order_id = str( uuid4())

    def execute(self):
        self.chef.prepare_pizza()
    


## Kitchen  ( Invoker )
from collections import deque
class Kitchen:

    def __init__(self):
        self.pending_orders = [] ## think of this as a queue


    def recieve_order(self, order:Order): ## here the order has to be an object (command pattern)
        print("Recieved new order -->", order.name)
        self.pending_orders.append(order)

    def cancel_order(self, order_id):

        self.pending_orders = [ order for order in self.pending_orders if order.order_id != order_id]
        print(f"order {order_id} was cancelled!!")
            
    def process_orders(self):
        ## pop from the queue --> assign the order for executed
        while self.pending_orders:
            current_order =  self.pending_orders.pop(0)
            current_order.execute()
```

```python
chef1 = Chef(name = "Sanjeev Kapoor")
chef2 = Chef(name = "Vikas Khanna")

kitchen = Kitchen()

# waiter has the order now!
burger_order = BurgerOrder(chef1)
pizza_order = PizzaOrder(chef2)
burger_order2 = BurgerOrder(chef1)


## waiter will go to kitchen ( kitchen will recieve the order and then invoke it to the chef)

kitchen.recieve_order(burger_order)
kitchen.recieve_order(burger_order2)
kitchen.recieve_order(pizza_order)


## kitchen will process the order
kitchen.cancel_order(burger_order2.order_id)
kitchen.process_orders()
```

# Chain of responsibility

    - Components:
        - handler interface
        - concrete handlers
        - client that can chain the handlers for a given request

 - naive way to handle chain of responsibility 
 - this is not good design (violating OCP)

class Handler:

    if ai_chatbot is able to resolve:
        ai_chatbot.resolve()
        request.close()
    
    elif human_chatbot is able to resolve:
        human_chatbot.resolve()
        request.close()

    elif ...

```python
# handler interface
class ComplaintHandler(ABC):

    def __init__(self):
        self._next = None

    # we should be able to decide the heirarchy
    def set_next(self, handler):
        self._next = handler
        return handler

    @abstractmethod
    def handle(self, request):
        if self._next:
            self._next.handle(request)


# concrete handlers
class AIChatbotHandler(ComplaintHandler):
    faq = { "cancel order" , "refund", "return", "delayed delivery" , "order status", "where is my order" }

    def handle(self, request):
        print("AI chatbot trying to handle the request!!")
        if request["issue"] in self.faq:
        
            print("chatbot can handle the request!!!")

        else:
            print("AI CHATBOT not able to handle the request!! passing to the next")
            super().handle(request)


class HumanChatbotHandler(ComplaintHandler):
    specific_problem = { "other concern", "reset password" }

    def handle(self, request):
        print("Human chatbot trying to handle the request!!")

        if request["issue"] in self.specific_problem:
            print("human chatbot can handle the request!!!")

        else:
            print("Human CHATBOT not able to handle the request!! passing to the next")
            super().handle(request)

class PhoneCallHandler(ComplaintHandler):
    critical_problem = { "blocked account related", "security related" }

    def handle(self, request):
        print("Phone Call Agent trying to handle the request!!")

        if request["issue"] in self.critical_problem or request["priority"] == 0:
            print("Phone Call Agent can handle the request!!!")

        else:
            print("We are not able to resolve.  We will escalate to the next. We respond in 2 days")
            super().handle(request)
               
```

```python
class SupportHandler(ABC):

    def __init__(self):
        self._next = None

    def set_next(self, handler):
        self._next = handler
        return handler

    @abstractmethod
    def handle(self, ticket):
        pass


class AutoResponder(SupportHandler):

    FAQ = {
        "reset password",
        "pricing",
        "hours"
    }

    def handle(self, ticket):

        if ticket["issue"].lower() in self.FAQ:
            print(f"AutoResponder handled: {ticket['issue']}")
            return

        print("AutoResponder cannot handle. Forwarding to next handler...")

        if self._next:
            self._next.handle(ticket)
        else:
            print("No handler available.")

class L1Support(SupportHandler):

    def handle(self, ticket):

        if ticket["priority"] <= 2:
            print(f"L1 handled: {ticket['issue']}")
            return

        print("L1 cannot handle. Forwarding to next handler...")

        if self._next:
            self._next.handle(ticket)
        else:
            print("No handler available.")

class L2Support(SupportHandler):

    def handle(self, ticket):

        if ticket["priority"] <= 4:
            print(f"L2 handled: {ticket['issue']}")
            return

        print("L2 cannot handle. Forwarding to next handler...")

        if self._next:
            self._next.handle(ticket)
        else:
            print("No handler available.")


class Engineering(SupportHandler):

    def handle(self, ticket):
        print(f"Engineering handled: {ticket['issue']}")


auto = AutoResponder()
l1 = L1Support()
l2 = L2Support()
engineering = Engineering()

auto.set_next(l1).set_next(l2).set_next(engineering)
```
