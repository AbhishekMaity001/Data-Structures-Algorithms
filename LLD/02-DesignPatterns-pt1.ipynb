# Simple Factory

```python
from abc import ABC, abstractmethod

class Notification(ABC):
    @abstractmethod
    def send():
        pass

    
###########################################################
class Email(Notification):
    def send(self):
        print("Sending email...")
        
class SMS(Notification):
    def send(self):
        print("Sending sms...")
        
class Whatsapp(Notification):
    def send(self):
        print("Sending whatsapp...")
###########################################################



class NotificationProvider:  # Factory creation. factory will be creating and giving us
    
    @staticmethod
    def provide_notification(notification_type):  # violating the OCP
        if notification_type=="email":
            return Email()
        
        if notification_type=="sms":
            return SMS()
        
        if notification_type=="whatsapp":
            return Whatsapp()
        
obj1 = NotificationProvider()
obj1.provide_notification("whatsapp").send()
```

# Factory Method

```python
###########################################################
class Notification:
    @abstractmethod
    def send(self, msg):
        pass
###########################################################


class Email(Notification):
    def send(self, msg):
        print(f"Sending email...{msg}")
        
class SMS(Notification):
    def send(self, msg):
        print(f"Sending sms...{msg}")
        
class Whatsapp(Notification):
    def send(self, msg):
        print(f"Sending whatsapp...{msg}")
        
        
###########################################################
class NotificationService(ABC):
    @abstractmethod
    def provide_notification() -> Notification:
        pass
    
    def notify(self, msg):
        notification = self.provide_notification()
        notification.send(msg)
###########################################################


class EmailService(NotificationService):
    def provide_notification(self) -> Notification:
        return Email()
class SMSService(NotificationService):
    def provide_notification(self) -> Notification:
        return SMS()
    
class WhatsappService(NotificationService):
    def provide_notification(self) -> Notification:
        return Whatsapp()
    
```

```python
obj1 = WhatsappService()
obj1.provide_notification().send("dsfs")
```

# Prototype pattern

    - Creational pattern that comes into picture when the object creation process is expensive and the object is created
      by cloning an existing object
    - we also use a prototype to create an object

# Copy, shallow copy and deep copy

```python
student1 = ["rajiv", "gopi", "sachin", [1,2,3]]
```

```python
id(student1)
```

```python
id(student1[0]), id(student1[1]), id(student1[2]), id(student1[3])
```

```python
import copy
student_shallow = copy.copy(student1)
```

```python
id(student_shallow)  # CREATED A NEW DABBA (container) BUT VALUES INSIDE THE CONTAINER ARE POINTING TO THE OLD OBJECT ONLY
```

```python
id(student_shallow[0]), id(student_shallow[1]), id(student_shallow[2]), id(student_shallow[3])
```

#### Appending to student1 but student_shallow is also getting affected

```python
student1[3].append(4)
```

```python
student1, student_shallow
```

```python
id(student1), id(student_shallow)  # different bcoz of immutable
```
