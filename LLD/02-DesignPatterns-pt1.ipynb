# Simple Factory

```python
from abc import ABC, abstractmethod

class Notification(ABC):
    @abstractmethod
    def send():
        pass

    
###########################################################
class Email(Notification):
    def send(self):
        print("Sending email...")
        
class SMS(Notification):
    def send(self):
        print("Sending sms...")
        
class Whatsapp(Notification):
    def send(self):
        print("Sending whatsapp...")
###########################################################



class NotificationProvider:  # Factory creation. factory will be creating and giving us
    
    @staticmethod
    def provide_notification(notification_type):  # violating the OCP
        if notification_type=="email":
            return Email()
        
        if notification_type=="sms":
            return SMS()
        
        if notification_type=="whatsapp":
            return Whatsapp()
        
obj1 = NotificationProvider()
obj1.provide_notification("whatsapp").send()
```

# Factory Method

```python
###########################################################
class Notification:
    @abstractmethod
    def send(self, msg):
        pass
###########################################################


class Email(Notification):
    def send(self, msg):
        print(f"Sending email...{msg}")
        
class SMS(Notification):
    def send(self, msg):
        print(f"Sending sms...{msg}")
        
class Whatsapp(Notification):
    def send(self, msg):
        print(f"Sending whatsapp...{msg}")
        
        
###########################################################
class NotificationService(ABC):
    @abstractmethod
    def provide_notification() -> Notification:
        pass
    
    def notify(self, msg):
        notification = self.provide_notification()
        notification.send(msg)
###########################################################


class EmailService(NotificationService):
    def provide_notification(self) -> Notification:
        return Email()
class SMSService(NotificationService):
    def provide_notification(self) -> Notification:
        return SMS()
    
class WhatsappService(NotificationService):
    def provide_notification(self) -> Notification:
        return Whatsapp()
    
```

```python
obj1 = WhatsappService()
obj1.provide_notification().send("dsfs")
```
